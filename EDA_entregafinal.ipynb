```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_theme(style="whitegrid", palette="deep")
plt.rcParams["figure.figsize"] = (10, 6)
pd.set_option("display.max_columns", None)
pd.set_option("display.max_rows", 100)
```

```python
df_eda = df_definitivo.copy()
```

```python
print("Dimensiones:", df_eda.shape)
display(df_eda.head())
```

```python
# Tipos de datos, memoria y valores no nulos
df_eda.info()

# Número de valores ausentes por columna
ausentes = df_eda.isna().sum().sort_values(ascending=False)
display(ausentes.to_frame("valores_ausentes"))

# Porcentaje de valores ausentes
porcentaje_ausentes = (df_eda.isna().mean() * 100).sort_values(ascending=False)
display(porcentaje_ausentes.to_frame("porcentaje_ausentes").round(2))

# Duplicados completos
print("Filas duplicadas:", df_eda.duplicated().sum())

# Columnas que parecen índices importados
columnas_indice = [c for c in df_eda.columns if c.lower().startswith("unnamed")]
print("Posibles columnas índice:", columnas_indice)
```

```python
# Trabajamos sobre la copia. Eliminamos únicamente columnas Unnamed
# que suelen ser índices exportados, comprobando primero cuáles existen.
df_eda = df_eda.drop(columns=columnas_indice, errors="ignore")

# En el CSV hay variables numéricas guardadas como texto con coma decimal
# y algunos valores etiquetados como 'Desconocido'.
columnas_coma = ["cons.price.idx", "cons.conf.idx", "euribor3m", "nr.employed"]

for col in columnas_coma:
    if col in df_eda.columns:
        df_eda[col] = (
            df_eda[col].astype("string")
            .str.replace(",", ".", regex=False)
            .replace({"Desconocido": pd.NA, "": pd.NA})
        )
        df_eda[col] = pd.to_numeric(df_eda[col], errors="coerce")

# Comprobamos el resultado
display(df_eda[columnas_coma].describe().T)
display(df_eda[columnas_coma].isna().sum().to_frame("ausentes"))
```

```python
# Estadísticos de variables numéricas
display(df_eda.describe().T)

# Resumen de variables categóricas
display(df_eda.describe(include=["object", "string"]).T)

# Frecuencias absolutas y relativas de las variables categóricas
columnas_categoricas = df_eda.select_dtypes(include=["object", "string", "category"]).columns

for col in columnas_categoricas:
    print(f"\nFrecuencias de {col}")
    frecuencias = df_eda[col].value_counts(dropna=False)
    porcentajes = df_eda[col].value_counts(normalize=True, dropna=False).mul(100)
    tabla = pd.DataFrame({
        "frecuencia": frecuencias,
        "porcentaje": porcentajes.round(2)
    })
    display(tabla.head(20))
```

```python
# Seleccionamos variables numéricas
columnas_numericas = df_eda.select_dtypes(include="number").columns.tolist()

# Histogramas: permiten observar forma, concentración y asimetría
df_eda[columnas_numericas].hist(
    bins=30, figsize=(16, 12), edgecolor="black"
)
plt.suptitle("Distribución de las variables numéricas", y=1.02)
plt.tight_layout()
plt.show()
```

```python
# Diagramas de caja para detectar posibles valores atípicos
for col in columnas_numericas:
    plt.figure(figsize=(9, 3))
    sns.boxplot(x=df_eda[col])
    plt.title(f"Diagrama de caja de {col}")
    plt.xlabel(col)
    plt.tight_layout()
    plt.show()
```

```python
# Gráficos de barras para las variables categóricas con cardinalidad moderada
for col in columnas_categoricas:
    n_categorias = df_eda[col].nunique(dropna=False)
    if n_categorias <= 20:
        orden = df_eda[col].value_counts().index
        plt.figure(figsize=(10, 5))
        sns.countplot(data=df_eda, y=col, order=orden)
        plt.title(f"Frecuencia de {col}")
        plt.xlabel("Número de registros")
        plt.ylabel(col)
        plt.tight_layout()
        plt.show()
```

```python
# Ingresos por nivel educativo: tamaño de muestra, media, mediana y dispersión
if {"education", "income"}.issubset(df_eda.columns):
    resumen_educacion = (
        df_eda.groupby("education", dropna=False)["income"]
        .agg(["count", "mean", "median", "std"])
        .sort_values("mean", ascending=False)
    )
    display(resumen_educacion)

    plt.figure(figsize=(11, 6))
    sns.boxplot(data=df_eda, x="income", y="education")
    plt.title("Distribución de ingresos según nivel educativo")
    plt.xlabel("Ingresos")
    plt.ylabel("Nivel educativo")
    plt.tight_layout()
    plt.show()
```

```python
# Número de hijos y adolescentes según educación
for variable in ["kidhome", "teenhome"]:
    if {"education", variable}.issubset(df_eda.columns):
        resumen = df_eda.groupby("education", dropna=False)[variable].agg(
            ["count", "mean", "median", "sum"]
        )
        display(resumen.sort_values("mean", ascending=False))

        plt.figure(figsize=(10, 5))
        sns.barplot(data=df_eda, x="education", y=variable, errorbar=None)
        plt.title(f"Media de {variable} según nivel educativo")
        plt.xticks(rotation=35, ha="right")
        plt.tight_layout()
        plt.show()
```

```python
if {"age", "income"}.issubset(df_eda.columns):
    datos = df_eda[["age", "income"]].dropna()
    datos = datos[datos["age"] > 0]

    # Ordenar los datos por edad
    datos = datos.sort_values("age")

    # Crear gráfico
    plt.figure(figsize=(10, 6))

    sns.scatterplot(
        data=datos,
        x="age",
        y="income",
        alpha=0.35
    )

    # Tendencia suavizada mediante media móvil
    tendencia = (
        datos.groupby("age")["income"]
        .mean()
        .rolling(window=5, center=True, min_periods=1)
        .mean()
    )

    plt.plot(
        tendencia.index,
        tendencia.values,
        color="red",
        linewidth=2
    )

    plt.title("Relación entre edad e ingresos")
    plt.xlabel("Edad")
    plt.ylabel("Ingresos")
    plt.tight_layout()
    plt.show()
```

```python
correlaciones = df_eda.select_dtypes(include="number").corr()

plt.figure(figsize=(13, 10))
sns.heatmap(
    correlaciones, cmap="coolwarm", center=0,
    annot=True, fmt=".2f", linewidths=0.5
)
plt.title("Matriz de correlación de variables numéricas")
plt.tight_layout()
plt.show()

# Pares con correlación absoluta más alta, sin contar la diagonal
correlaciones_abs = correlaciones.abs().where(
    np.triu(np.ones(correlaciones.shape), k=1).astype(bool)
)
display(
    correlaciones_abs.stack()
    .sort_values(ascending=False)
    .to_frame("correlacion_absoluta")
    .head(15)
)
```

```python
if "y" in df_eda.columns:
    display(df_eda["y"].value_counts(dropna=False).to_frame("frecuencia"))
    display((df_eda["y"].value_counts(normalize=True, dropna=False) * 100)
            .round(2).to_frame("porcentaje"))

    plt.figure(figsize=(7, 4))
    sns.countplot(data=df_eda, x="y", order=df_eda["y"].value_counts().index)
    plt.title("Distribución del resultado de la campaña")
    plt.xlabel("Resultado (y)")
    plt.ylabel("Número de registros")
    plt.tight_layout()
    plt.show()

    # Tasa de respuesta por categorías relevantes
    for variable in ["education", "job", "marital", "contact"]:
        if variable in df_eda.columns:
            tasa = df_eda.groupby(variable, dropna=False)["y"].apply(
                lambda s: (s.astype("string").str.lower() == "yes").mean() * 100
            ).sort_values(ascending=False)
            display(tasa.to_frame("porcentaje_yes").round(2))
```

```python
# En los datos observados, age contiene valores 0.
# Revisamos cuántos hay y qué proporción representan; no los sustituimos sin
# conocer el significado que tenían en el proceso de limpieza.
if "age" in df_eda.columns:
    n_cero = (df_eda["age"] == 0).sum()
    print(f"Registros con age == 0: {n_cero:,}")
    print(f"Porcentaje: {n_cero / len(df_eda) * 100:.2f}%")

# Comprobación de etiquetas 'Desconocido' que puedan seguir presentes
for col in df_eda.select_dtypes(include=["object", "string"]).columns:
    n_desc = df_eda[col].astype("string").str.strip().str.lower().eq("desconocido").sum()
    if n_desc:
        print(f"{col}: {n_desc} valores 'Desconocido'")
```
